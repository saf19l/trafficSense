```python
import cv2
import os

videos = [
    "23712-337108764_medium.mp4",
    "4531753-hd_1080_1920_25fps.mp4",
    "5927708-hd_1080_1920_30fps.mp4"
]

output_folder = "traffic_frames"
os.makedirs(output_folder, exist_ok=True)

saved = 0

for video_path in videos:
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    print(f"Processing: {video_path}")

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if frame_count % 10 == 0:
            filename = f"{output_folder}/frame_{saved:04d}.jpg"
            cv2.imwrite(filename, frame)
            saved += 1
        frame_count += 1

    cap.release()
    print(f"Done! Frames saved so far: {saved}")

print(f"\nTotal frames saved: {saved}")
print(f"Check the '{output_folder}' folder in your miniproject directory!")
```

```python
from ultralytics import YOLO
import cv2
import os

# Load YOLO model
model = YOLO("yolov8n.pt")

# Vehicle class IDs
vehicle_classes = [2, 3, 5, 7]

def calculate_density(vehicle_count):
    if vehicle_count <= 5:
        return "Low"
    elif vehicle_count <= 15:
        return "Medium"
    else:
        return "High"

def get_signal_time(density):
    if density == "Low":
        return 15
    elif density == "Medium":
        return 30
    else:
        return 45

# Run on all frames
frames_folder = "traffic_frames"
all_frames = sorted(os.listdir(frames_folder))

print("Frame | Vehicles | Density | Green Light Time")
print("-" * 50)

total_vehicles = 0
low = medium = high = 0

for frame_name in all_frames:
    frame_path = os.path.join(frames_folder, frame_name)
    frame = cv2.imread(frame_path)
    
    results = model(frame, verbose=False)
    
    vehicle_count = 0
    for result in results:
        for box in result.boxes:
            if int(box.cls) in vehicle_classes:
                vehicle_count += 1
    
    density = calculate_density(vehicle_count)
    signal_time = get_signal_time(density)
    total_vehicles += vehicle_count
    
    if density == "Low": low += 1
    elif density == "Medium": medium += 1
    else: high += 1
    
    print(f"{frame_name} | {vehicle_count:8d} | {density:7s} | {signal_time} seconds")

print("\n" + "=" * 50)
print(f"Total frames processed: {len(all_frames)}")
print(f"Total vehicles detected: {total_vehicles}")
print(f"Low density frames: {low}")
print(f"Medium density frames: {medium}")
print(f"High density frames: {high}")
print(f"Average vehicles per frame: {total_vehicles/len(all_frames):.1f}")
```
